## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

**Question:** which pages should a content editor review first, because their impressions are likely to fall?

**Decision it improves:** how an editor spends limited review time. A team can open only a limited number of pages each week, so the order matters.

**Who acts:** a content editor or SEO lead, who reads the ranked list, opens each page, and decides whether to change it.

**Cost of a wrong call:** a false alarm wastes review time (I assume about 30 minutes per page), and a change made to a page that did not need one can introduce a problem. Missing a page that really is declining means the problem is noticed later. Because of this, every recommendation goes through a person and the list is decision-support, not an automatic action.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/iankitthakur/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)
print("pages in the starter slice:", len(df), "| columns:", df.shape[1])

down = df["trend_direction"].str.lower().eq("down")
print("pages whose impressions went down:", int(down.sum()), f"({round(down.mean() * 100, 1)}%)")
print("median change in impressions among pages with a value (%):", round(df["trend_pct"].median(), 1))
df["trend_direction"].value_counts()
```

Three numbers from the starter slice (observed): it holds 30,000 pages; 54.2% of them had their impressions go down between the previous and the last 30 days; and among pages with a measured change, the median change was about −33.5%. More than half the pages are declining, so an editor who can review only a few dozen pages needs a sensible order for them. That makes the lane worth the next seven weeks.

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

**What my work can say:**
- Observed: patterns in this sample of pages, for example how many pages declined and how clicks and position relate.
- Measured: how a rule or a model ranks pages compared with a baseline, on the same split, with the base rate shown.
- Directional: signals that tend to go together with decline in this data, without a claim of cause.
- Decision-support: a ranked list for a human to review, with reason codes.

**What it will never say:**
- That an edit or refresh caused or will cause traffic to recover.
- That I know how Google ranks pages, or that I can predict Google.
- That results from one month and one sample apply to every site.
- Anything about a named client, domain or private query.
